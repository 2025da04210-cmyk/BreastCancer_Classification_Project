```python
# ==========================================================
# STEP 2 : UNDERSTAND THE DATASET
# ==========================================================

import pandas as pd

# Load Dataset
df = pd.read_csv("Documents\ML_2\data.csv")

print("="*60)
print("DATASET INFORMATION")
print("="*60)

# First 5 Rows
print("\nFirst 5 Rows")
print(df.head())

# Shape
print("\nDataset Shape")
print("Rows :", df.shape[0])
print("Columns :", df.shape[1])

# Feature Names
print("\nFeature Names")
print(df.columns.tolist())

# Dataset Information
print("\nDataset Info")
df.info()

# Data Types
print("\nData Types")
print(df.dtypes)

# Missing Values
print("\nMissing Values")
print(df.isnull().sum())

# Duplicate Records
print("\nDuplicate Records :", df.duplicated().sum())

# Target Variable
print("\nTarget Variable : diagnosis")

print("\nTarget Distribution")
print(df["diagnosis"].value_counts())

print("\nTarget Percentage")
print(df["diagnosis"].value_counts(normalize=True) * 100)

# Numerical Columns
numerical_columns = df.select_dtypes(include=["int64","float64"]).columns

print("\nNumber of Numerical Columns :", len(numerical_columns))
print(numerical_columns.tolist())

# Categorical Columns
categorical_columns = df.select_dtypes(include=["object"]).columns

print("\nNumber of Categorical Columns :", len(categorical_columns))
print(categorical_columns.tolist())

# Statistical Summary
print("\nStatistical Summary")
print(df.describe())

print("\nSTEP 2 COMPLETED SUCCESSFULLY")
```

```python
# ==========================================================
# STEP 3 : DATA PREPROCESSING
# ==========================================================

import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler
# ----------------------------------------------------------
# Remove Unnecessary Columns
# ----------------------------------------------------------

df.drop(columns=["id", "Unnamed: 32"], inplace=True)

print("\nDataset Shape After Removing Columns")
print(df.shape)

# ----------------------------------------------------------
# Missing Values
# ----------------------------------------------------------

print("\nMissing Values Before Cleaning")
print(df.isnull().sum())

# Fill Missing Values (if any)

for col in df.select_dtypes(include=["float64","int64"]).columns:
    df[col] = df[col].fillna(df[col].median())

for col in df.select_dtypes(include=["object"]).columns:
    df[col] = df[col].fillna(df[col].mode()[0])

print("\nMissing Values After Cleaning")
print(df.isnull().sum())

# ----------------------------------------------------------
# Encode Target Variable
# ----------------------------------------------------------

label_encoder = LabelEncoder()

df["diagnosis"] = label_encoder.fit_transform(df["diagnosis"])

print("\nDiagnosis Encoding")
print("Benign (B) -> 0")
print("Malignant (M) -> 1")

print(df["diagnosis"].value_counts())

# ----------------------------------------------------------
# Separate Features and Target
# ----------------------------------------------------------

X = df.drop("diagnosis", axis=1)

y = df["diagnosis"]

print("\nFeature Matrix Shape :", X.shape)
print("Target Shape :", y.shape)

# ----------------------------------------------------------
# Train-Test Split
# ----------------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining Samples :", len(X_train))
print("Testing Samples :", len(X_test))

# ----------------------------------------------------------
# Feature Scaling
# ----------------------------------------------------------

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)

print("\nFeature Scaling Completed")

# ----------------------------------------------------------
# Final Shapes
# ----------------------------------------------------------

print("\nFinal Dataset Shapes")

print("X_train :", X_train_scaled.shape)
print("X_test  :", X_test_scaled.shape)
print("y_train :", y_train.shape)
print("y_test  :", y_test.shape)

print("\nSTEP 3 COMPLETED SUCCESSFULLY")
```

```python
# ==========================================================
# STEP 4 : MACHINE LEARNING CLASSIFICATION MODELS
# ==========================================================
# Classification Models
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.ensemble import RandomForestClassifier

# Evaluation Metrics
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    matthews_corrcoef
)
# ==========================================================
# DEFINE MODELS
# ==========================================================

models = {
    "Logistic Regression": LogisticRegression(random_state=42, max_iter=5000),

    "Decision Tree": DecisionTreeClassifier(random_state=42),

    "K-Nearest Neighbors": KNeighborsClassifier(n_neighbors=5),

    "Gaussian Naive Bayes": GaussianNB(),

    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        random_state=42
    )
}

# ==========================================================
# MODEL TRAINING & EVALUATION
# ==========================================================

results = []

print("="*100)
print("MODEL EVALUATION")
print("="*100)

for name, model in models.items():

    # Train model
    model.fit(X_train, y_train)

    # Prediction
    y_pred = model.predict(X_test)

    # Probability for AUC
    y_prob = model.predict_proba(X_test)[:, 1]

    # Metrics
    accuracy = accuracy_score(y_test, y_pred)

    auc = roc_auc_score(y_test, y_prob)

    precision = precision_score(y_test, y_pred)

    recall = recall_score(y_test, y_pred)

    f1 = f1_score(y_test, y_pred)

    mcc = matthews_corrcoef(y_test, y_pred)

    # Store Results
    results.append([
        name,
        accuracy,
        auc,
        precision,
        recall,
        f1,
        mcc
    ])

    # Print Results
    print("\n", "="*60)
    print(name)
    print("="*60)

    print("Accuracy :", round(accuracy,4))
    print("AUC Score :", round(auc,4))
    print("Precision :", round(precision,4))
    print("Recall :", round(recall,4))
    print("F1 Score :", round(f1,4))
    print("MCC Score :", round(mcc,4))

# ==========================================================
# FINAL COMPARISON TABLE
# ==========================================================

results_df = pd.DataFrame(
    results,
    columns=[
        "Model",
        "Accuracy",
        "AUC Score",
        "Precision",
        "Recall",
        "F1 Score",
        "MCC Score"
    ]
)

print("\n")
print("="*100)
print("FINAL MODEL COMPARISON")
print("="*100)

print(results_df)

# Save Results
results_df.to_csv("Model_Comparison_Results.csv", index=False)

print("\nResults saved successfully as 'Model_Comparison_Results.csv'")
```

```python
# ==========================================================
# STEP 6 : CONFUSION MATRIX & CLASSIFICATION REPORT
# ==========================================================

from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report

for name, model in models.items():

    print("\n")
    print("="*80)
    print(name)
    print("="*80)

    # Prediction
    y_pred = model.predict(X_test)

    # Confusion Matrix
    cm = confusion_matrix(y_test, y_pred)

    print("\nConfusion Matrix")
    print(cm)

    # Classification Report
    print("\nClassification Report")
    print(classification_report(y_test, y_pred))
```

```python
# ==========================================================
# STEP 7 : MODEL COMPARISON
# ==========================================================

print("\n")
print("="*100)
print("MODEL COMPARISON")
print("="*100)

print(results_df)

print("\nHighest Accuracy")
print(results_df.loc[results_df["Accuracy"].idxmax()])

print("\nHighest AUC Score")
print(results_df.loc[results_df["AUC Score"].idxmax()])

print("\nHighest Precision")
print(results_df.loc[results_df["Precision"].idxmax()])

print("\nHighest Recall")
print(results_df.loc[results_df["Recall"].idxmax()])

print("\nHighest F1 Score")
print(results_df.loc[results_df["F1 Score"].idxmax()])

print("\nHighest MCC Score")
print(results_df.loc[results_df["MCC Score"].idxmax()])

# ----------------------------------------------------------
# Best Overall Model
# ----------------------------------------------------------

best_model = results_df.iloc[results_df["Accuracy"].idxmax()]

print("\n")
print("="*100)
print("BEST MODEL")
print("="*100)

print("Best Model :", best_model["Model"])
print("Accuracy :", round(best_model["Accuracy"],4))
print("AUC Score :", round(best_model["AUC Score"],4))
print("Precision :", round(best_model["Precision"],4))
print("Recall :", round(best_model["Recall"],4))
print("F1 Score :", round(best_model["F1 Score"],4))
print("MCC Score :", round(best_model["MCC Score"],4))
```
